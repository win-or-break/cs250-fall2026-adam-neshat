```python
def slices_needed(people, per_person):
    return people * per_person


def pizzas_to_order(slices, per_pizza):
    pizzas, leftover = divmod(slices, per_pizza)
    if leftover > 0:        # a partial pizza still means buying a whole one
        pizzas += 1
    return pizzas


def plan_party(people, per_person, per_pizza, price):
    slices = slices_needed(people, per_person)
    pizzas = pizzas_to_order(slices, per_pizza)
    leftover_slices = pizzas * per_pizza - slices
    cost_per_person = round(pizzas * price / people, 2)
    return (pizzas, leftover_slices, cost_per_person)
```
